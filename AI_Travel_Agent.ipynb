```python
from google import genai

client = genai.Client(api_key="")
```

```python
print("Available Models and their Supported Actions:")
for model in client.models.list():
    print(f"- {model.name}: {model.supported_actions}")
```

```python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Hello"
)

print(response.text)
```

```python
def travel_agent(user_input):

    prompt = f"""
    You are an AI Travel Agent.

    Think step-by-step before answering.

    Follow these reasoning steps:

    1. Destination Analysis
    - Identify destination
    - Identify travel type
    - Identify group type

    2. Budget Evaluation
    - Detect budget level
    - Estimate transport cost
    - Estimate hotel cost

    3. Travel Duration Planning
    - Plan activities day-wise

    4. Hotel Recommendation
    - Suggest suitable hotels
    - Explain why selected

    5. Transportation Planning
    - Suggest transport
    - Estimate transport cost

    6. Activity Suggestions
    - Recommend tourist places
    - Recommend foods
    - Recommend activities

    7. Final Itinerary
    - Create a clean day-wise travel plan

    User Request:
    {user_input}

    Important:
    Explain why each recommendation was selected.
    Keep the answer organized and easy to understand.
    """

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python
query = "I want a 3-day low-budget trip to Cox's Bazar with my friends."

result = travel_agent(query)

print(result)
```

```python
import gradio as gr

interface = gr.Interface(
    fn=travel_agent,

    inputs=gr.Textbox(
        lines=4,
        placeholder="Enter your travel plan here...",
        label="Travel Request"
    ),

    outputs=gr.Textbox(
        lines=25,
        label="AI Travel Plan"
    ),

    title="AI Travel Agent",
    description="Travel Planner using Gemini API",

    theme="soft"
)

interface.launch()
```
